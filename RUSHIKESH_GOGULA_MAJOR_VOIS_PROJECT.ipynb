However, differences in averages alone do not confirm that season is statistically responsible for the variation. Therefore, ANOVA is used to test the statistical significance of seasonal differences.

### **7.3 Analysis of Variance (ANOVA)**

One-way ANOVA is used to determine whether the mean agricultural performance differs significantly across Kharif, Rabi and Zaid seasons.

The null hypothesis assumes that the mean values are equal across all seasons.

The alternative hypothesis assumes that at least one season has a significantly different mean.

A significance level of **5% (α = 0.05)** is used.

````python
# Perform one-way ANOVA to test whether average yield differs significantly across seasons

from scipy.stats import f_oneway

kharif_yield = df[df['Season'] == 'Kharif']['Yield_Tonnes_Ha'].dropna()
rabi_yield = df[df['Season'] == 'Rabi']['Yield_Tonnes_Ha'].dropna()
zaid_yield = df[df['Season'] == 'Zaid']['Yield_Tonnes_Ha'].dropna()

f_statistic, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("ANOVA Test for Yield Across Seasons")
print("-----------------------------------")
print("F-statistic:", round(f_statistic, 4))
print("P-value:", round(p_value, 6))

if p_value < 0.05:
    print("\nResult: The difference in mean yield across seasons is statistically significant.")
else:
    print("\nResult: The difference in mean yield across seasons is not statistically significant.")
````

### **Interpretation**

The ANOVA test gives an F-statistic of approximately **1.458** and a p-value of approximately **0.233**.

Since the p-value is greater than **0.05**, the null hypothesis is not rejected.

Therefore, the differences in average yield between Kharif, Rabi and Zaid are **not statistically significant at the 5% significance level**.

Although Kharif has the highest average yield and Zaid has the lowest average yield, the statistical test indicates that the observed seasonal differences in yield are not strong enough to conclude that season alone creates a significant difference in yield.

This suggests that other factors such as crop type, irrigation, resource usage and economic conditions may also contribute to agricultural performance.

### **7.4 ANOVA of Agricultural Profit Across Seasons**

ANOVA is also applied to `Profit_INR` to determine whether average agricultural profitability differs significantly across seasons.

This is useful because agricultural performance is not limited to production and yield; economic profitability is also an important outcome.

````python
# Perform one-way ANOVA to test whether average profit differs significantly across seasons

kharif_profit = df[df['Season'] == 'Kharif']['Profit_INR']
rabi_profit = df[df['Season'] == 'Rabi']['Profit_INR']
zaid_profit = df[df['Season'] == 'Zaid']['Profit_INR']

f_statistic_profit, p_value_profit = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("ANOVA Test for Profit Across Seasons")
print("------------------------------------")
print("F-statistic:", round(f_statistic_profit, 4))
print("P-value:", p_value_profit)

if p_value_profit < 0.05:
    print("\nResult: Seasonal differences in average profit are statistically significant.")
else:
    print("\nResult: Seasonal differences in average profit are not statistically significant.")
````

### **Interpretation**

The ANOVA test produces an F-statistic of approximately **34.292** with a p-value far below **0.05**.

Therefore, the null hypothesis is rejected.

This indicates that there is a **statistically significant difference in average agricultural profit across seasons**.

The result is important because although seasonal differences in average yield were not statistically significant, seasonal differences in **profitability were significant**.

This suggests that economic performance is influenced by factors beyond yield alone, such as production levels, market prices, input costs, resource usage and other seasonal conditions.

### **7.5 Correlation Analysis**

Correlation analysis is used to measure the strength and direction of relationships between agricultural variables.

The Pearson correlation coefficient ranges from **-1 to +1**.

A value close to +1 indicates a strong positive relationship, a value close to -1 indicates a strong negative relationship, and a value close to 0 indicates a weak linear relationship.

````python
# Calculate Pearson correlation between important variables and agricultural yield

correlation_variables = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Profit_INR',
    'Seed_Quality_Score'
]

correlation_with_yield = (
    df[correlation_variables + ['Yield_Tonnes_Ha']]
    .corr()['Yield_Tonnes_Ha']
    .drop('Yield_Tonnes_Ha')
    .sort_values(ascending=False)
)

correlation_with_yield.round(4)
````

### **Interpretation**

The correlation analysis shows that **Water Efficiency** has the strongest positive relationship with agricultural yield, with a correlation coefficient of approximately **0.92**.

Water Used also shows a moderate positive relationship with yield, with a correlation coefficient of approximately **0.39**.

Profit has a moderate positive relationship with yield, with a correlation coefficient of approximately **0.49**.

In comparison, rainfall, temperature, humidity, soil moisture and fertilizer usage show very weak linear relationships with yield in this dataset.

These results indicate that agricultural yield is strongly associated with water efficiency in the available data. However, correlation does not prove causation, so these relationships should not be interpreted as direct cause-and-effect relationships.
